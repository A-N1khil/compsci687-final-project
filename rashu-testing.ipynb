```python
from mdp.mountaincar.mountaincar import MountainCarMDP
from algorithms.actor_critic.actor_critic import ActorCriticAgent

mdp = MountainCarMDP(seed=42)
state_dim = mdp.state_space.shape[0]
action_dim = mdp.action_space.n

agent = ActorCriticAgent(mdp, state_dim, action_dim)
rewards = agent.train(num_episodes=400, max_steps=200, log_interval=10)
print(rewards)
print("Best reward:", max(rewards))


```

```python
import matplotlib.pyplot as plt
plt.plot(rewards)
plt.title("MountainCar Actor-Critic Reward Progress")
plt.xlabel("Episode")
plt.ylabel("Shaped Reward")
plt.show()
```

```python
env = MountainCarMDP()

state = env.reset()
for i in range(10):
    action = env.get_action_space().sample()
    next_state, reward, done, info = env.step(state, action)
    print(f"Step {i}: Action={action}, Reward={reward}, Done={done}")
    state = next_state
    if done:
        print("Episode ended early!")
        break
```

```python
from mdp.mountaincar.mountaincar import MountainCarMDP
from algorithms.actor_critic.actor_critic import ActorCriticAgent

mdp = MountainCarMDP(seed=42)
state_dim = mdp.state_space.shape[0]
action_dim = mdp.action_space.n

agent = ActorCriticAgent(
    mdp,
    state_dim,
    action_dim,
    use_reward_shaping=True,   # 🔴 important for MountainCar
)

rewards = agent.train(num_episodes=400, max_steps=200, log_interval=10)
print("Best raw reward:", max(rewards))

```

```python
from mdp.mountaincar.mountaincar_continous import MountainCarContinuousMDP
from algorithms.actor_critic.actor_critic_continous import ContinuousActorCriticAgent

mdp = MountainCarContinuousMDP(seed=42)
state_dim = mdp.get_state_space().shape[0]
action_dim = mdp.get_action_space().shape[0]

agent = ContinuousActorCriticAgent(
    mdp,
    state_dim,
    action_dim,
    gamma=0.99,
    lr=3e-4,
    entropy_coef=1e-3,
)

rewards = agent.train(num_episodes=2000, max_steps=1000, log_interval=10)

print("Best raw reward:", max(rewards))

```
